# Task 1: While Loop & Control Statements -Number Guessing Game


## ★ Generate a random number between 1 and 10 that the user has to guess.Import random and use randint function.

### 🟢 A while loop repeats a block of code as long as a condition is true.
### 🟢 It is useful when the number of repetitions is not fixed.
### 🟢 The loop checks the condition before every iteration.
### 🟢 Break is used to stop the loop immediately, even if the condition is still true.
### 🟢 Continue skips the remaining code inside the loop for the current iteration and moves directly to the next iteration of the loop.

&nbsp;


## 🟢 In a number‑guessing game:

### 🔸while keeps asking for guesses
### 🔸Break stops the game when the guess is correct
### 🔸Continue handles invalid inputs without stopping the game

&nbsp;


```python
import random

# Generate random number between 1 and 10
secret_number = random.randint(1, 10)

print("Welcome to the Number Guessing Game!")
print("Guess a number between 1 and 10.")

while True:
    guess = int(input("\nEnter your guess: "))
    if guess == secret_number:

        print("Correct! Congratulations! You guessed the correct number..")
        break


    elif guess < 1 or guess > 10:
        print("Your guess is out of range. Please guess a number between 1 and 10.")
        continue


    else:
        print("Too low. Try again..")

else:
    print("Game Over!")
```

## Task 2:
## For Loop - Multiplication Table Generator

### The program generates a multiplication table from 1 to 10 for any number entered by the user.

### 🔹It uses a for loop and the range() function to repeat calculations

### 🔹A for loop is used to iterate through numbers 1 to 10.
### 🔹The range(1, 11) function generates numbers from 1 up to 10.
### 🔹 In each loop cycle, the program multiplies the user’s number with the current value of i.

```python
# Task 2: Multiplication Table using For Loop

num = int(input("Enter the number for which you want the multiplication table: "))

print(f"\nMultiplication Table for {num}:\n")

for i in range(1, 10+1):
    result = num * i
    print(f" {num} x {i} = {result}")
```

## Task 3: Function- BMI Calculator

## <u>Formula Used in Python</u>
### 🔸 BMI is calculated using the formula:
###  🔸BMI = weight / (height × height)
### 🔸 Weight is taken in kilograms (kg).
### 🔸 Height is taken in meters (m).

## 1. Defining the Function
### 🔹 A function named calculate_bmi(weight, height) is created.
### 🔹 The function takes two inputs:
####    - weight (in kilograms)
####    - height (in meters)
### 🔹Inside the function, BMI is calculated using the formula:
####    - BMI = weight / (height × height)
### 🔹 The function returns the calculated BMI value to the main program.

```python
# Task 3: BMI Calculator using Function

def calculate_bmi(weight, height):
    return weight / (height ** 2)

weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in meters: "))

bmi = calculate_bmi(weight, height)

print(f"Your BMI is: {bmi:.2f}")
```
